```python
import os
import torch
# --- Configuration ---
# Languages for the MLCommons/ml_spoken_words dataset
os.environ['LANG'] = 'en --lang es'  # Example: English and Spanish
os.environ['EPOCHS'] = '5'
os.environ['BATCH_SIZE'] = '16'
os.environ['LR'] = '1e-4'
os.environ['CNN_DIM'] = '64'
os.environ['GRU_HIDDEN'] = '128'
os.environ['GRU_LAYERS'] = '2'
os.environ['OUTPUT_DIM'] = '128' # Lower output dim for a smaller student
os.environ['CROP_SECONDS'] = '2.0'
os.environ['MLFLOW_URI'] = 'http://192.168.1.200:5000'
os.environ['EXPERIMENT'] = 'Kaggle_TinyHuBERT'
os.environ['PLOT_INTERVAL'] = '1'

# The device should automatically be 'cuda' if a GPU is available on Kaggle
os.environ['DEVICE'] = 'cuda' if torch.cuda.is_available() else 'cpu'

# Optional: Set a checkpoint path to resume training (e.g., os.environ['RESUME'] = 'last_checkpoint.pt')
os.environ['RESUME'] = ''

print("Configuration parameters set as environment variables.")
```

```python
# download tinyhubert.py
# URL to the raw content of the Gist file: tinyhubert.py
GIST_RAW_URL="https://gist.githubusercontent.com/JarbasAl/8135a4c755c5b9234c713a3d497cabfb/raw/253e35f538fb40b8359f1530b7f924b843472519/wakehubert_distill.py"

# Download the file using wget
!wget -O tinyhubert.py $GIST_RAW_URL

print("tinyhubert.py downloaded successfully.")
!ls -l tinyhubert.py
```

```python
!python --version
!python tinyhubert.py \
    --lang $LANG \
    --epochs $EPOCHS \
    --batch-size $BATCH_SIZE \
    --lr $LR \
    --cnn-dim $CNN_DIM \
    --gru-hidden $GRU_HIDDEN \
    --gru-layers $GRU_LAYERS \
    --output-dim $OUTPUT_DIM \
    --crop-seconds $CROP_SECONDS \
    --mlflow-uri $MLFLOW_URI \
    --experiment $EXPERIMENT \
    --device $DEVICE \
    --plot-interval $PLOT_INTERVAL \
    --resume $RESUME
```
